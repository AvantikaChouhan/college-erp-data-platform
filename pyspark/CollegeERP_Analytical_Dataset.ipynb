```python
students = spark.read.csv(
    "/Volumes/workspace/default/collegeerp_data/students.csv",
    header=True,
    inferSchema=True
)

attendance = spark.read.csv(
    "/Volumes/workspace/default/collegeerp_data/attendance.csv",
    header=True,
    inferSchema=True
)

results = spark.read.csv(
    "/Volumes/workspace/default/collegeerp_data/results.csv",
    header=True,
    inferSchema=True
)

fees = spark.read.csv(
    "/Volumes/workspace/default/collegeerp_data/fees.csv",
    header=True,
    inferSchema=True
)

placements = spark.read.csv(
    "/Volumes/workspace/default/collegeerp_data/placements.csv",
    header=True,
    inferSchema=True
)

print("Students:", students.count())
print("Attendance:", attendance.count())
print("Results:", results.count())
print("Fees:", fees.count())
print("Placements:", placements.count())
```

```python
from pyspark.sql.functions import avg, count, col, round, when, sum

# =========================
# 1. Student Master
# =========================

student_master = students.select(
    "roll_no",
    "student_name",
    "department",
    "semester"
)


# =========================
# 2. Attendance Summary
# =========================

attendance_summary = attendance.groupBy(
    "roll_no"
).agg(
    round(
        avg("attendance_percentage"), 2
    ).alias("avg_attendance")
)


# =========================
# 3. Results Summary
# =========================

results_summary = results.groupBy(
    "roll_no"
).agg(
    round(
        avg("marks"), 2
    ).alias("avg_marks"),
    
    count("subject_code").alias(
        "total_subjects"
    )
)


# =========================
# 4. Fees
# =========================

fees_selected = fees.select(
    "roll_no",
    "total_fee",
    "amount_paid",
    "pending_amount",
    "payment_status"
)


# =========================
# 5. Placements
# =========================

placements_selected = placements.select(
    "roll_no",
    "company",
    "package_lpa",
    "placement_status"
)


# =========================
# 6. Final Analytical Dataset
# =========================

final_dataset = student_master \
    .join(
        attendance_summary,
        on="roll_no",
        how="left"
    ) \
    .join(
        results_summary,
        on="roll_no",
        how="left"
    ) \
    .join(
        fees_selected,
        on="roll_no",
        how="left"
    ) \
    .join(
        placements_selected,
        on="roll_no",
        how="left"
    )


# =========================
# 7. Handle Missing Placement Data
# =========================

final_dataset = final_dataset.withColumn(
    "company",
    when(
        col("company").isNull(),
        "N/A"
    ).otherwise(col("company"))
)

final_dataset = final_dataset.withColumn(
    "package_lpa",
    when(
        col("package_lpa").isNull(),
        0.0
    ).otherwise(col("package_lpa"))
)

final_dataset = final_dataset.withColumn(
    "placement_status",
    when(
        col("placement_status").isNull(),
        "Not Placed"
    ).otherwise(col("placement_status"))
)


# =========================
# 8. Display Final Dataset
# =========================

print("Total Records:", final_dataset.count())
print("Total Columns:", len(final_dataset.columns))

display(final_dataset)
```

```python
from pyspark.sql.functions import (
    avg, count, sum, max, min, round,
    col, when
)

kpi_df = spark.sql("""
SELECT
    COUNT(*) AS total_students,

    ROUND(AVG(avg_attendance), 2) AS avg_attendance,

    ROUND(AVG(avg_marks), 2) AS avg_marks,

    SUM(pending_amount) AS total_pending_fees,

    SUM(amount_paid) AS total_fees_collected,

    COUNT(
        CASE
            WHEN avg_attendance < 75 THEN 1
        END
    ) AS students_below_75_attendance,

    COUNT(
        CASE
            WHEN avg_marks < 50 THEN 1
        END
    ) AS students_below_50_marks,

    COUNT(
        CASE
            WHEN placement_status = 'Placed' THEN 1
        END
    ) AS placed_students,

    ROUND(
        100.0 *
        COUNT(
            CASE
                WHEN placement_status = 'Placed' THEN 1
            END
        ) / COUNT(*),
        2
    ) AS placement_rate,

    ROUND(
        AVG(
            CASE
                WHEN package_lpa > 0 THEN package_lpa
            END
        ),
        2
    ) AS avg_package_lpa,

    MAX(package_lpa) AS highest_package_lpa

FROM collegeerp_student_analytics
""")

display(kpi_df)
```

```python
insight_df = spark.sql("""
SELECT
    COUNT(*) AS total_students,

    -- Academic Risk
    SUM(
        CASE
            WHEN avg_attendance < 75
             AND avg_marks < 50
            THEN 1 ELSE 0
        END
    ) AS academic_risk_students,

    -- Attendance Risk
    SUM(
        CASE
            WHEN avg_attendance < 75
            THEN 1 ELSE 0
        END
    ) AS attendance_risk_students,

    -- Fee Risk
    SUM(
        CASE
            WHEN pending_amount > 0
             AND avg_marks < 50
            THEN 1 ELSE 0
        END
    ) AS fee_academic_risk_students,

    -- High performers who are not placed
    SUM(
        CASE
            WHEN avg_marks >= 70
             AND placement_status = 'Not Placed'
            THEN 1 ELSE 0
        END
    ) AS high_performers_unplaced,

    -- Strong attendance + academics
    SUM(
        CASE
            WHEN avg_attendance >= 75
             AND avg_marks >= 70
            THEN 1 ELSE 0
        END
    ) AS academically_strong_students,

    -- Students with pending fees
    SUM(
        CASE
            WHEN pending_amount > 0
            THEN 1 ELSE 0
        END
    ) AS students_with_pending_fees

FROM collegeerp_student_analytics
""")

display(insight_df)
```

```python
department_insights = spark.sql("""
SELECT
    department,

    COUNT(*) AS total_students,

    ROUND(AVG(avg_attendance), 2)
        AS avg_attendance,

    ROUND(AVG(avg_marks), 2)
        AS avg_marks,

    SUM(pending_amount)
        AS pending_fees,

    COUNT(
        CASE
            WHEN placement_status = 'Placed'
            THEN 1
        END
    ) AS placed_students,

    ROUND(
        100.0 *
        COUNT(
            CASE
                WHEN placement_status = 'Placed'
                THEN 1
            END
        ) / COUNT(*),
        2
    ) AS placement_rate,

    ROUND(
        AVG(
            CASE
                WHEN package_lpa > 0
                THEN package_lpa
            END
        ),
        2
    ) AS avg_package_lpa

FROM collegeerp_student_analytics

GROUP BY department

ORDER BY department
""")

display(department_insights)
```

```python
attendance_performance = spark.sql("""
SELECT
    CASE
        WHEN avg_attendance < 60 THEN 'Below 60%'
        WHEN avg_attendance < 75 THEN '60% - 74%'
        WHEN avg_attendance < 85 THEN '75% - 84%'
        ELSE '85%+'
    END AS attendance_band,

    COUNT(*) AS students,

    ROUND(AVG(avg_marks), 2) AS average_marks

FROM collegeerp_student_analytics

GROUP BY
    CASE
        WHEN avg_attendance < 60 THEN 'Below 60%'
        WHEN avg_attendance < 75 THEN '60% - 74%'
        WHEN avg_attendance < 85 THEN '75% - 84%'
        ELSE '85%+'
    END

ORDER BY
    CASE
        WHEN attendance_band = 'Below 60%' THEN 1
        WHEN attendance_band = '60% - 74%' THEN 2
        WHEN attendance_band = '75% - 84%' THEN 3
        ELSE 4
    END
""")

display(attendance_performance)
```

```python
student_segments = spark.sql("""
SELECT
    CASE
        WHEN avg_attendance >= 75
         AND avg_marks >= 70
         AND placement_status = 'Placed'
            THEN 'Placement Ready'

        WHEN avg_attendance >= 75
         AND avg_marks >= 70
         AND placement_status = 'Not Placed'
            THEN 'High Performer - Unplaced'

        WHEN avg_attendance < 75
         AND avg_marks < 50
            THEN 'Academic Risk'

        WHEN pending_amount > 0
         AND avg_marks < 50
            THEN 'Fee + Academic Risk'

        ELSE 'Regular'
    END AS student_segment,

    COUNT(*) AS students

FROM collegeerp_student_analytics

GROUP BY
    CASE
        WHEN avg_attendance >= 75
         AND avg_marks >= 70
         AND placement_status = 'Placed'
            THEN 'Placement Ready'

        WHEN avg_attendance >= 75
         AND avg_marks >= 70
         AND placement_status = 'Not Placed'
            THEN 'High Performer - Unplaced'

        WHEN avg_attendance < 75
         AND avg_marks < 50
            THEN 'Academic Risk'

        WHEN pending_amount > 0
         AND avg_marks < 50
            THEN 'Fee + Academic Risk'

        ELSE 'Regular'
    END

ORDER BY students DESC
""")

display(student_segments)
```

```python
department_360 = spark.sql("""
SELECT
    department,

    COUNT(*) AS total_students,

    ROUND(AVG(avg_attendance), 2) AS avg_attendance,

    ROUND(AVG(avg_marks), 2) AS avg_marks,

    SUM(pending_amount) AS pending_fees,

    COUNT(
        CASE
            WHEN pending_amount > 0 THEN 1
        END
    ) AS students_with_pending_fees,

    COUNT(
        CASE
            WHEN avg_attendance < 75
             AND avg_marks < 50
            THEN 1
        END
    ) AS academic_risk_students,

    COUNT(
        CASE
            WHEN placement_status = 'Placed'
            THEN 1
        END
    ) AS placed_students,

    ROUND(
        100.0 *
        COUNT(
            CASE
                WHEN placement_status = 'Placed'
                THEN 1
            END
        ) / COUNT(*),
        2
    ) AS placement_rate,

    ROUND(
        AVG(
            CASE
                WHEN package_lpa > 0
                THEN package_lpa
            END
        ),
        2
    ) AS avg_package_lpa

FROM collegeerp_student_analytics

GROUP BY department

ORDER BY department
""")

display(department_360)
```

```python
placement_insights = spark.sql("""
SELECT
    department,

    COUNT(*) AS total_students,

    COUNT(
        CASE
            WHEN placement_status = 'Placed'
            THEN 1
        END
    ) AS placed_students,

    COUNT(
        CASE
            WHEN placement_status = 'Not Placed'
            THEN 1
        END
    ) AS unplaced_students,

    ROUND(
        100.0 *
        COUNT(
            CASE
                WHEN placement_status = 'Placed'
                THEN 1
            END
        ) / COUNT(*),
        2
    ) AS placement_rate,

    ROUND(
        AVG(
            CASE
                WHEN placement_status = 'Placed'
                THEN avg_marks
            END
        ),
        2
    ) AS avg_marks_placed,

    ROUND(
        AVG(
            CASE
                WHEN placement_status = 'Not Placed'
                THEN avg_marks
            END
        ),
        2
    ) AS avg_marks_unplaced,

    ROUND(
        AVG(
            CASE
                WHEN placement_status = 'Placed'
                THEN avg_attendance
            END
        ),
        2
    ) AS avg_attendance_placed,

    ROUND(
        AVG(
            CASE
                WHEN placement_status = 'Not Placed'
                THEN avg_attendance
            END
        ),
        2
    ) AS avg_attendance_unplaced,

    ROUND(
        AVG(
            CASE
                WHEN package_lpa > 0
                THEN package_lpa
            END
        ),
        2
    ) AS avg_package_lpa,

    MAX(package_lpa) AS highest_package_lpa

FROM collegeerp_student_analytics

GROUP BY department

ORDER BY placement_rate DESC
""")

display(placement_insights)
```

```python
fee_academic_insights = spark.sql("""
SELECT
    CASE
        WHEN pending_amount > 0
         AND avg_marks < 50
         AND avg_attendance < 75
            THEN 'Critical Risk'

        WHEN pending_amount > 0
         AND (avg_marks < 50 OR avg_attendance < 75)
            THEN 'Moderate Risk'

        WHEN pending_amount > 0
            THEN 'Fee Pending Only'

        WHEN avg_marks < 50
         OR avg_attendance < 75
            THEN 'Academic Risk Only'

        ELSE 'Low Risk'
    END AS risk_category,

    COUNT(*) AS students,

    ROUND(AVG(avg_marks), 2) AS avg_marks,

    ROUND(AVG(avg_attendance), 2) AS avg_attendance,

    SUM(pending_amount) AS total_pending_fees

FROM collegeerp_student_analytics

GROUP BY
    CASE
        WHEN pending_amount > 0
         AND avg_marks < 50
         AND avg_attendance < 75
            THEN 'Critical Risk'

        WHEN pending_amount > 0
         AND (avg_marks < 50 OR avg_attendance < 75)
            THEN 'Moderate Risk'

        WHEN pending_amount > 0
            THEN 'Fee Pending Only'

        WHEN avg_marks < 50
         OR avg_attendance < 75
            THEN 'Academic Risk Only'

        ELSE 'Low Risk'
    END

ORDER BY students DESC
""")

display(fee_academic_insights)
```

```python
semester_insights = spark.sql("""
SELECT
    semester,
    COUNT(*) AS total_students,

    ROUND(AVG(avg_attendance), 2) AS avg_attendance,

    ROUND(AVG(avg_marks), 2) AS avg_marks,

    SUM(pending_amount) AS pending_fees,

    COUNT(
        CASE
            WHEN placement_status = 'Placed'
            THEN 1
        END
    ) AS placed_students,

    ROUND(
        100.0 *
        COUNT(
            CASE
                WHEN placement_status = 'Placed'
                THEN 1
            END
        ) / COUNT(*),
        2
    ) AS placement_rate

FROM collegeerp_student_analytics

GROUP BY semester

ORDER BY semester
""")

display(semester_insights)
```

```python
top_packages = spark.sql("""
SELECT
    roll_no,
    student_name,
    department,
    semester,
    ROUND(avg_marks, 2) AS avg_marks,
    ROUND(avg_attendance, 2) AS avg_attendance,
    company,
    ROUND(package_lpa, 2) AS package_lpa

FROM collegeerp_student_analytics

WHERE placement_status = 'Placed'

ORDER BY package_lpa DESC

LIMIT 10
""")

display(top_packages)
```

```python
company_insights = spark.sql("""
SELECT
    company,
    COUNT(*) AS placed_students,
    ROUND(AVG(package_lpa), 2) AS avg_package_lpa,
    ROUND(MAX(package_lpa), 2) AS highest_package_lpa

FROM collegeerp_student_analytics

WHERE placement_status = 'Placed'
  AND company <> 'N/A'

GROUP BY company

ORDER BY placed_students DESC, avg_package_lpa DESC
""")

display(company_insights)
```
